```python
# remove duplicate rows
print("Before:", df.shape)
df = df.drop_duplicates(subset='customer_id')
print("After:", df.shape)
```

```python
# fic the nagetive age
df['age'] = df['age'].abs()
```

```python
# categorical column case fix
cat_cols = ['gender', 'marital_status', 'employment_type', 'loan_type', 'home_ownership', 'default']

for col in cat_cols:
    df[col] = df[col].astype(str).str.strip().str.lower()

df['gender'].unique() 
```

```python
# standerize default column
df['default'] = df['default'].replace({'y': 'yes', 'n': 'no'})
df['default'] = df['default'].str.capitalize() 
df['default'].value_counts()
```

```python
# set date format
df['application_date'] = pd.to_datetime(df['application_date'], format='mixed', dayfirst=False)
df['application_date'].head(10)
```

```python
# fill the missing values
# Numeric columns
for col in ['monthly_income', 'credit_score', 'debt_to_income_ratio', 'num_dependents']:
    df[col] = df[col].fillna(df[col].median())

# Categorical 
df['home_ownership'] = df['home_ownership'].fillna('unknown')

df.isnull().sum()  
```
